### Fetch HatchMD motes using the API 

### Imports

```python
import os, requests, re
from IPython.display import display, Markdown
from pathlib import Path
```

#### Constants

```python
BASE = "https://api.hackmd.io/v1"
H = {"Authorization": f"Bearer {os.environ['HACKMD_API_TOKEN']}"}
CHILD = re.compile(r"\]\(/([A-Za-z0-9_-]{16,})\)")
out = Path("hackmd_notes"); out.mkdir(exist_ok=True)
```

#### Functions

```python
def get_note(note_id):
    r = requests.get(f"{BASE}/notes/{note_id}", headers=H)
    r.raise_for_status()
    return r.json()
```

#### Main Call site and Use Regex to get links

```python
out = Path("hackmd_notes"); out.mkdir(exist_ok=True)
index = get_note("DSF-FPTFRvygvriuR9GJEA")
ids = list(dict.fromkeys(CHILD.findall(index["content"])))
print(f"{len(ids)} child notes found")
```

#### Save all notes locally

```python
for nid in ids:
    try:
        n = get_note(nid)
    except requests.HTTPError as e:
        print(f"  skip {nid}: {e.response.status_code}")
        continue
    slug = re.sub(r"[^\w-]+", "-", n["title"] or nid).strip("-").lower()[:60]
    (out / f"{slug}.md").write_text(n["content"])
    print(f"  {slug}.md")
```
